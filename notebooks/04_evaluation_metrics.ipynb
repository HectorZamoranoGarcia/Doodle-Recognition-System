# Notebook 4: Linear Separability Test

In this notebook, it will be mathematically verified whether the *MyDoodles* dataset is **linearly separable**.

The Perceptron algorithm will be utilized on the entire dataset (without dividing into training and testing). According to the **Perceptron Convergence Theorem**, if the data are linearly separable, the algorithm must be able to reach **0 errors** within a finite number of steps.

If the error does not reach 0 after many epochs, empirical proof will be established that non-linear models (such as Neural Networks) are required for this problem.

```python
import numpy as np
import os

# 1. Load Data
X_PATH = "../data/doodles_X.npy"
y_PATH = "../data/doodles_y.npy"

if os.path.exists(X_PATH):
    X = np.load(X_PATH)
    y = np.load(y_PATH).astype(int)
    print(f"Data loaded: {X.shape[0]} samples of {X.shape[1]} dimensions.")
else:
    raise FileNotFoundError("Files are missing in ../data/")

# Split is NOT performed. Everything is used for the separability test.
N, D = X.shape
num_classes = len(np.unique(y))

print(f"Objective: Fit {num_classes} classes using the {N} available samples.")
```

```python
print("Initiating Linear Separability Test...")
print("-" * 40)
print(f"{'Epoch':^10} {'Total Errors':^20} {'Status':^10}")
print("-" * 40)

# --- CONFIGURATION ---
max_epochs = 1000       # Many opportunities are provided
alpha = 1.0             # Standard learning rate
W = np.zeros((1+D, num_classes)) # Weights are initialized to 0

is_separable = False

# --- LOOP 1: EPOCHS (Attempts) ---
for epoch in range(1, max_epochs + 1):
    errors = 0

    # --- LOOP 2: SAMPLES (One by one) ---
    for n in range(N):
        # 1. Prepare input (with bias)
        xn = np.insert(X[n], 0, 1)
        target = y[n]

        # 2. Prediction (Dot product)
        # Scores are calculated for all classes
        scores = xn @ W
        prediction = np.argmax(scores)

        # 3. Update (If it fails)
        if prediction != target:
            # Penalty to the wrongly predicted class
            W[:, prediction] -= alpha * xn
            # Reward to the correct class
            W[:, target] += alpha * xn
            errors += 1

    # --- PRINT TABLE ROW ---
    # The first 10 are always shown, then every 50, and always if it reaches 0
    if epoch <= 10 or epoch % 50 == 0 or errors == 0:
        status = "YES" if errors == 0 else "NO"
        print(f"{epoch:^10d} {errors:^20d} {status:^10}")

    # STOPPING CRITERION: If errors are 0, the process is finished.
    if errors == 0:
        is_separable = True
        break

print("-" * 40)

if is_separable:
    print(f"CONCLUSION: The data ARE linearly separable!")
    print(f"  The algorithm converged at epoch {epoch}.")
else:
    print(f"CONCLUSION: The data are NOT linearly separable.")
    print(f"  After {max_epochs} epochs, the model could not drop below {errors} errors.")
```
