# 🧪 Notebook 4: Test de Separabilidad Lineal

En este cuaderno vamos a comprobar matemáticamente si nuestro dataset *MyDoodles* es **linealmente separable**.

Utilizaremos el algoritmo del Perceptrón sobre la totalidad de los datos (sin dividir en entrenamiento y test). Según el **Teorema de Convergencia del Perceptrón**, si los datos son linealmente separables, el algoritmo debe ser capaz de llegar a **0 errores** en un número finito de pasos.

Si tras muchas épocas el error no llega a 0, tendremos la demostración empírica de que necesitamos modelos no lineales (como Redes Neuronales) para este problema.

```python
import numpy as np
import os

# 1. Cargar Datos
X_PATH = "../data/doodles_X.npy"
y_PATH = "../data/doodles_y.npy"

if os.path.exists(X_PATH):
    X = np.load(X_PATH)
    y = np.load(y_PATH).astype(int)
    print(f"✅ Datos cargados: {X.shape[0]} muestras de {X.shape[1]} dimensiones.")
else:
    raise FileNotFoundError("❌ Faltan los archivos en ../data/")

# NO hacemos split. Usamos todo para el test de separabilidad.
N, D = X.shape
num_clases = len(np.unique(y))

print(f"🎯 Objetivo: Ajustar {num_clases} clases usando las {N} muestras disponibles.")
```

```python
print("Iniciando Test de Separabilidad Lineal...")
print("-" * 40)
print(f"{'Época':^10} {'Errores Totales':^20} {'Estado':^10}")
print("-" * 40)

# --- CONFIGURACIÓN ---
max_epochs = 1000       # Damos muchas oportunidades
alpha = 1.0             # Tasa de aprendizaje estándar
W = np.zeros((1+D, num_clases)) # Inicializamos pesos a 0

es_separable = False

# --- BUCLE 1: ÉPOCAS (Intentos) ---
for epoch in range(1, max_epochs + 1):
    errors = 0

    # --- BUCLE 2: MUESTRAS (Uno a uno) ---
    for n in range(N):
        # 1. Preparar entrada (con sesgo)
        xn = np.insert(X[n], 0, 1)
        target = y[n]

        # 2. Predicción (Producto punto)
        # Calculamos la puntuación para todas las clases
        scores = xn @ W
        prediction = np.argmax(scores)

        # 3. Actualización (Si falla)
        if prediction != target:
            # Castigo a la clase predicha erróneamente
            W[:, prediction] -= alpha * xn
            # Premio a la clase correcta
            W[:, target] += alpha * xn
            errors += 1

    # --- IMPRIMIR FILA DE LA TABLA ---
    # Mostramos siempre las primeras 10, luego cada 50, y siempre si llega a 0
    if epoch <= 10 or epoch % 50 == 0 or errors == 0:
        estado = "✅" if errors == 0 else "❌"
        print(f"{epoch:^10d} {errors:^20d} {estado:^10}")

    # CRITERIO DE PARADA: Si errores es 0, hemos terminado.
    if errors == 0:
        es_separable = True
        break

print("-" * 40)

if es_separable:
    print(f"🎉 ¡CONCLUSIÓN: Los datos SON linealmente separables!")
    print(f"   El algoritmo convergió en la época {epoch}.")
else:
    print(f"⚠️ CONCLUSIÓN: Los datos NO son linealmente separables.")
    print(f"   Tras {max_epochs} épocas, el modelo no pudo bajar de {errors} errores.")
```
